```python
step = np.array([
    [5000, 6200, 7100],
    [8000, 7500, 9000],
    [4500, 5100, 4800],
    [9000, 8500, 9500]
])

total_step=np.sum(step)
print(total_step)
mean_step=np.mean(step)
print(mean_step)
max_step=np.max(step)
min_step=np.min(step)
print(max_step)
print(min_step)
total_step_day=np.sum(step,axis=0)
total_step_user=np.sum(step,axis=1)
print(total_step_day)
print(total_step_user)
position = np.unravel_index(np.argmax(step), step.shape)
print(position)
```

```python
X = np.array([
    [6, 70, 3],
    [5, 50, 6],
    [8, 80, 2],
    [4, 30, 8]
])

y = np.array([40, 65, 30, 85])

print(X.shape)
print(X.ndim)
XT=X.T
print(XT)
XT_X = X.T @ X
print(XT_X)

inverse = np.linalg.inv(XT_X)
print(inverse)

beta = inverse @ X.T @ y


print("Sleep:", beta[0])
print("Activity:", beta[1])
print("Stress:", beta[2])


new_user = np.array([5, 40, 7])
prediction = new_user @ beta

print("Q4(g) Predicted assistance score:", prediction)
```
