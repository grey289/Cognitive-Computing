### Q3 - Random numbers

```python
import random

random.seed(1024170249)
random_numbers = [random.randint(100, 900) for _ in range(100)]
print(random_numbers)
```

```python
odd_numbers = [n for n in random_numbers if n % 2 != 0]
print(len(odd_numbers))
```

```python
even_numbers = [n for n in random_numbers if n % 2 == 0]
print(len(even_numbers))
```

```python
def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False
    return True

prime_numbers = [n for n in random_numbers if is_prime(n)]
print(len(prime_numbers))
print(prime_numbers)
```

```python
most_common = max(set(random_numbers), key=random_numbers.count)
print(most_common, random_numbers.count(most_common))
```

### Q4 - Sets

```python
digits_8 = digits[:8]

A = {d * 7 for d in digits_8}
B = {d * 9 for d in digits_8}

print(A)
print(B)
```

```python
print(A.union(B))
```

```python
print(A.intersection(B))
```

```python
print(A.difference(B))
print(B.difference(A))
# difference() only gives one-sided diff, symmetric_difference() gives both combined
```

```python
print(A.symmetric_difference(B))
```

```python
print(A.issubset(B))
print(B.issuperset(A))
```

```python
try:
    X = int(input("value to remove from A: "))
except Exception:
    X = next(iter(A))

A.discard(X)
print(A)
# discard wont throw an error if X isnt in the set, remove() would
```

### Q5 - Dictionaries

```python
my_dict = {
    "name": "Sahu",
    "roll_no": "1024170249",
    "branch": "COPC",
    "age": 20,
    "city": "Delhi"
}
print(my_dict)
```

```python
my_dict["location"] = my_dict.pop("city")
print(my_dict)
```

```python
my_dict["cgpa"] = 9.77
print(my_dict)
```

```python
my_dict["age"] += 1
print(my_dict)
```

```python
dict_copy1 = my_dict.copy()
dict_copy2 = my_dict.copy()

removed_value = dict_copy1.pop("branch")
print(dict_copy1, removed_value)

del dict_copy2["branch"]
print(dict_copy2)
# pop() gives back the value it removed, del doesnt return anything
```

```python
for key, value in my_dict.items():
    print(key, "->", value)
```

```python
if "email" in my_dict:
    print(my_dict["email"])
else:
    print("no email key")
```

```python
friend_dict = {
    "name": "Rohan Verma",
    "roll_no": "1024170999",
    "branch": "CSE",
    "age": 21,
    "city": "Delhi"
}

merged_dict = {**my_dict, **friend_dict}
print(merged_dict)
# friend_dict values overwrite my_dict ones on common keys since its unpacked second
```

```python
string_valued = {k: v for k, v in my_dict.items() if isinstance(v, str)}
print(string_valued)
```
