```python
import numpy as np

arr = np.array([1, 2, 3, 4, 5])

arr = arr + 2
print(arr)

arr = arr * 3
print(arr)

arr = arr / 2
print(arr)
```

```python
arr=np.flip(arr)
print(arr)
```

```python
x = np.array([1,2,3,4,5,1,2,1,1,1])

values,counts=np.unique(x,return_counts=True)

max_count=np.max(counts)

most_freq=values[counts==max_count]

print("Most frequent element is ",most_freq)

for val in most_freq:
  print("Indices ",np.where(x==val)[0])
```

```python
y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3])

values, counts = np.unique(y, return_counts=True)

most_count = np.max(counts)
most_freq = values[counts == most_count]

print("Most frequent element(s):", most_freq)

for val in most_freq:
    print("Indices:", np.where(y == val)[0])
```

```python
sahu=np.linspace(10,100,25)
print("array ",sahu)
print("dimensions ",sahu.ndim)
print("shape ",sahu.shape)
print("total elements ",sahu.size)
print("Data type : ", sahu.dtype)
print("Total Bytes : ",sahu.nbytes)

print("using reshape")
print(sahu.reshape(1,25))

print("using T")
print(sahu.reshape(1,25).T)
```

```python
ucs420_sahu=np.array([
    [10,20,30],
    [40,50,60],
    [70,80,90],
    [15,20,35]
])
print("original matrix")
print(ucs420_sahu)

print("mean ",np.mean(ucs420_sahu))
print("Median ",np.median(ucs420_sahu))
print("Max ",np.max(ucs420_sahu))
print("Min ",np.min(ucs420_sahu))
print("Unique elements ",np.unique(ucs420_sahu))

reshaped_ucs420_sahu=ucs420_sahu.reshape(4,3)
print(reshaped_ucs420_sahu)

resize_ucs420_sahu=np.resize(ucs420_sahu,(2,3))
print(resize_ucs420_sahu)
```
