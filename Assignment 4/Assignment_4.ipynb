```python
import pandas as pd

fixed_entries = [
    {
        "question": "what is the annual fee",
        "answer": "The annual fee is Rs 500.",
        "keywords": "fee cost price charge",
        "category": "billing"
    },
    {
        "question": "how to reset password",
        "answer": "Go to Settings > Reset Password.",
        "keywords": "password reset login",
        "category": "account"
    },
    {
        "question": "what are your working hours",
        "answer": "We are open 9 AM to 5 PM.",
        "keywords": "hours timing open time",
        "category": "general"
    },
    {
        "question": "how can i pay the fee",
        "answer": "You can pay via UPI, card, or net banking.",
        "keywords": "pay payment upi fee",
        "category": "billing"
    }
]

personalized_entries = [
    {
        "question": "how do i update my email address",
        "answer": "Go to Settings > Account > Update Email.",
        "keywords": "email update account",
        "category": "account"
    },
    {
        "question": "how do i get a fee receipt",
        "answer": "You can download your fee receipt from the payment section.",
        "keywords": "receipt fee payment",
        "category": "billing"
    }
]

entries = fixed_entries + personalized_entries
df = pd.DataFrame(entries)

print("Q1:")
print(df)


def score_query(query, df):
    query_words = set(query.lower().split())
    results = []

    for index, row in df.iterrows():
        text = row["question"].lower() + " " + row["keywords"].lower()
        text_words = set(text.split())
        score = len(query_words.intersection(text_words))

        if score > 0:
            results.append({
                "index": index,
                "question": row["question"],
                "answer": row["answer"],
                "category": row["category"],
                "score": score
            })

    results = sorted(results, key=lambda x: x["score"], reverse=True)

    return pd.DataFrame(results)


query = input("\nQ2: Enter your query: ")
result = score_query(query, df)

print("\nMatching FAQs:")
print(result)


def same_category(category_name, df):
    return df[df["category"].str.lower() == category_name.lower()]


print("\nQ3:")
print(same_category("account", df))


selected_index = 0

print("\nQ4:")
print(df.loc[selected_index])

new_keyword = input("Enter a new keyword: ")

df.loc[selected_index, "keywords"] = (
    df.loc[selected_index, "keywords"] + " " + new_keyword
)

print("\nUpdated FAQ:")
print(df.loc[selected_index])

filename = "49_faq_data.csv"
df.to_csv(filename, index=False)

print("Data saved as:", filename)


print("\nQ5:")
category_counts = df.groupby("category").size()
print(category_counts)


def score_query_with_ties(query, df):
    query_words = set(query.lower().split())
    results = []

    for index, row in df.iterrows():
        text = row["question"].lower() + " " + row["keywords"].lower()
        text_words = set(text.split())
        score = len(query_words.intersection(text_words))

        if score > 0:
            results.append({
                "index": index,
                "question": row["question"],
                "answer": row["answer"],
                "category": row["category"],
                "score": score
            })

    if len(results) == 0:
        print("No matching FAQ found.")
        return

    results = sorted(results, key=lambda x: x["score"], reverse=True)
    result_df = pd.DataFrame(results)

    highest_score = result_df["score"].max()

    best_matches = result_df[
        result_df["score"] == highest_score
    ]

    print("\nHighest score:", highest_score)

    if len(best_matches) > 1:
        print("Tie detected:")
        print(best_matches)
    else:
        print("Best match:")
        print(best_matches)


print("\nQ6 - Tie:")
score_query_with_ties("fee", df)

print("\nQ6 - No tie:")
score_query_with_ties("password", df)
```
